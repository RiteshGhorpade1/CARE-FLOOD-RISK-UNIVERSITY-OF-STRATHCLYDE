```python
import geopandas as gpd
import matplotlib.pyplot as plt

# Load both files - update paths to where you unzipped
pva = gpd.read_file("FRM_Local_Plan_Districts_v2_GPKG/Data/LPD.gpkg")
frm = gpd.read_file("GeoPackage/Data/PVAs v3.gpkg")

# Align Data
print("PVA shape:", pva.shape)
print("PVA CRS:", pva.crs)
print("PVA columns:", pva.columns.tolist())
print()
print("FRM shape:", frm.shape)
print("FRM CRS:", frm.crs)
print("FRM columns:", frm.columns.tolist())

# Plot both
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
pva.plot(ax=axes[0], color='steelblue', edgecolor='white', linewidth=0.3)
axes[0].set_title("FRM Local Plan Districts — Scotland")
axes[0].axis('off')

frm.plot(ax=axes[1], color='coral', edgecolor='white', linewidth=0.3)
axes[1].set_title("Potentially Vulnerable Areas — Scotland")
axes[1].axis('off')

plt.tight_layout()
plt.savefig("VS_Downloads/sepa_overview.png", dpi=150)
plt.show()
```
